# Rag pipeline from data ingestion to vector db pipeline

```python
import os
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

## Read all the pdf inside the dir

```python
def load_pdf_files(pdf_files_path):
    all_documents = []
    pdf_dir = Path(pdf_files_path)

    pdf_files = list(pdf_dir.glob("*.pdf"))

    print(f"found {len(pdf_files)} files in the file directory")

    for file in pdf_files:
        print(f"Processing file: {file}")
        try:
            loader = PyMuPDFLoader(file)
            documents = loader.load()

            all_documents.extend(documents)
            print(f"Pages in {file}: {len(documents)}")
            print(f"Loaded all document to: {all_documents}")
        
        except Exception as e:
            print(f"Error: {e}")
    print(f"Added all the documents {len(all_documents)}")
    return all_documents


```

```python
from pathlib import Path
path_directory = "../data/pdf_files"

all_pdf_documents = load_pdf_files(path_directory)

```

```python
# text splitting get into chunks

def split_documents(documents, chunk_size=500, chunk_overlap=200):
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap = chunk_overlap,
        length_function=len,
        separators = ["\n\n","\n"," ",""]
    )

    split_doc = text_splitter.split_documents(documents)
    print(f"Split {len(documents)} into {len(split_doc)} chunks")

    # Show sample chunk
    if split_doc:
        print("Sample split documents:\n")
        print("Document metadata:",split_doc[0].metadata)
        print(f"\nDocument page content: {split_doc[0].page_content[:200]}...")
    
    return split_doc
```

```python
split_documents(all_pdf_documents)
```
